## Question 1 — Heat produced by pushing a rock

A rock is pushed 20 meters over a rough horizontal concrete floor and then left without any more movement. The force applied to move it was constant at 500 N. How much heat is produced? Assume that no wear happened.

```python
# Given values
F = 500       # N
d = 20        # m

# Work done = force × distance
W = F * d

# Since there is no wear and the rock starts/ends at rest,
# the work is converted into heat.
heat = W

print(f"Work done = {W:.0f} J")
print(f"Heat produced = {heat:.0f} J = {heat/1000:.1f} kJ")
```

**Answer:** 10,000 J = **10 kJ** of heat is produced.

## Question 2 — Gunpowder needed to accelerate a bullet

A bullet for a gun has a mass of 6.75 g. Gunpowder can provide a speed of up to 500 m/s. If the gunpowder liberates 3 kJ/g, how many grams of gunpowder are needed to accelerate the bullet to 500 m/s if half of the energy is converted into linear movement of the bullet?

```python
# Given values
bullet_mass_g = 6.75       # g
bullet_mass_kg = bullet_mass_g / 1000
v = 500                      # m/s
energy_per_g = 3             # kJ/g
efficiency = 0.5             # 50%

# Kinetic energy of the bullet
KE = 0.5 * bullet_mass_kg * v**2

# Only 50% of gunpowder energy becomes kinetic energy
required_gunpowder_energy = KE / efficiency

# Gunpowder mass
gunpowder_mass_g = required_gunpowder_energy / energy_per_g

print(f"Bullet kinetic energy = {KE:.5f} J = {KE/1000:.5f} kJ")
print(f"Required gunpowder energy = {required_gunpowder_energy:.5f} kJ")
print(f"Gunpowder needed = {gunpowder_mass_g:.5f} g")
```

**Answer:** Approximately **0.563 g of gunpowder** is needed.

## Question 3 — Rotational kinetic energy of a cylinder

A cylinder with a mass of 3 kg and radius 5 cm is rotating at 2000 rpm. The moment of inertia of a solid cylinder is $I = \frac{1}{2}mR^2$. What is its rotational kinetic energy?

```python
# Given values
m = 3            # kg
R = 0.05         # m
rpm = 2000       # revolutions per minute

# Moment of inertia of a solid cylinder
I = 0.5 * m * R**2

# Convert rpm to radians per second
import math
omega = rpm * 2 * math.pi / 60

# Rotational kinetic energy
KE_rot = 0.5 * I * omega**2

print(f"Moment of inertia = {I:.6f} kg·m²")
print(f"Angular velocity = {omega:.5f} rad/s")
print(f"Rotational kinetic energy = {KE_rot:.2f} J")
```

**Answer:** Approximately **82.25 J** of rotational kinetic energy.

## Question 4 — Power needed to fill a rooftop water tank

A 10 m³ water tank near the roof of a skyscraper is located with its centre of mass 82 m above the ground. What power is needed to fill it completely in one hour, assuming no losses? The density of water is 1 g/cm³ and gravitational acceleration is 9.82 m/s².

```python
# Given values
V = 10                 # m³
h = 82                 # m
density = 1000         # kg/m³ (1 g/cm³)
g = 9.82               # m/s²
t = 3600               # s (1 hour)

# Mass of water
mass = density * V

# Gravitational potential energy
E = mass * g * h

# Required power
P = E / t

print(f"Mass of water = {mass:.0f} kg")
print(f"Energy required = {E:.0f} J")
print(f"Power required = {P:.2f} W = {P/1000:.3f} kW")
```

**Answer:** Approximately **2.24 kW** of power is needed.

## Question 5 — Carnot efficiency of a biomass power station

A biomass-fired power station receives heat from combustion at 350 °C. The exit temperature of the turbine is 50 °C. Calculate the Carnot efficiency.

```python
# Given temperatures
T_hot_C = 350       # °C
T_cold_C = 50       # °C

# Convert Celsius to Kelvin
T_hot_K = T_hot_C + 273.15
T_cold_K = T_cold_C + 273.15

# Carnot efficiency
efficiency = 1 - T_cold_K / T_hot_K

print(f"Hot temperature = {T_hot_K:.2f} K")
print(f"Cold temperature = {T_cold_K:.2f} K")
print(f"Carnot efficiency = {efficiency:.4f} = {efficiency*100:.2f}%")
```

**Answer:** The Carnot efficiency is approximately **48.2%**.
