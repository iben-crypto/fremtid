## Q3. Thermodynamics and electrochemistry i.

Q3. Thermodynamics and electrochemistry i.
Which statement about Gibbs free energy is most correct

[| A: The Gibbs free energy of a process quantifies the energy that can be utilised
for mechanical or electrical work without violation of the second law of
thermodynamics

B: The Gibbs free energy of a process quantifies the total energy liberated from
a process

C: The Gibbs free energy is similar to internal energy but with an adjustment
for the necessary pressure-volume work

D: The Gibbs free energy determines reaction kinetics and thus how much
work can be produced

E: The Gibbs free energy of a process is always negative

**Your answer:**


<details>
<summary>Show answer</summary>

The total energy liberated from or consumed by a chemical process at constant pressure is
the enthalpy of reaction. If this is to be converted into work, either mechanical or electrical
work, one needs to consider the second law of thermodynamics stating that entropy cannot
be destroyed. If the process leads to decreased entropy, typically by a net reduction of the
amount of gas molecules, then that amount of entropy must be generated by using part of
the energy for generation of heat. The rest of the energy from the process can be used freely
for anything and is therefore named free energy or Gibbs free energy.

</details>
